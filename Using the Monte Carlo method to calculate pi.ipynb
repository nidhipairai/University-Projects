I will be using the Monte Carlo method, which uses random numbers, to solve a problem. In this case, I will use this algorithm to approximate the value of $\pi$.

```python
# Importing libraries
import matplotlib.pyplot as plt
import numpy as np
```

## Visualising the approximation by using diagrams
The fraction of the total points within the quarter circle will be approximately equal to $\pi/4$

```python
# Create a figure
def shapes():
    """Draw a unit square and inscribed quarter-circle."""
    # Draw quarter circle with unit radius
    theta_circ = np.linspace(0,np.pi/2,1001)
    x_circ     = np.cos(theta_circ)
    y_circ     = np.sin(theta_circ)
    plt.plot(x_circ,y_circ,'-b')
    plt.gca().set_aspect('equal') # set aspect ratio to prevent shape distortion

    # Draw unit square
    plt.plot([0,1,1,0,0],[0,0,1,1,0],'b')

plt.title("Fig 1. Semicircle plotted in a unit square"); # Semicolon ; on last line suppresses unwanted text output
    
# Draw unit square and inscribed quarter-circle
shapes()
```

```python
# Create figure with unit square and inscribed quarter-circle
plt.figure()
shapes()

# Generate random arrays of x and y coordinates within unit square
n_points = 1000
x_points = np.random.rand(n_points)
y_points = np.random.rand(n_points)

# Plot generated points
plt.plot(x_points,y_points,'.');

plt.title("Random numbers plotted on Fig.1");
```

#### By counting the number of points within unit distance of the origin, and dividing it by the total number of points, we get a ratio that should be approximately $\pi/4$. So by taking this ratio and multiplying it by 4, we get an approximation of $\pi$.

```python
n_inside = 0
for i in range(n_points):
    x = np.random.rand()
    y = np.random.rand()
    if (x**2 + y**2) < 1:
        n_inside += 1

pi_approx = 4 * n_inside / n_points
print(f"Of {n_points} points, {n_inside} are inside the quarter circle.")
print(f"Estimate of pi = {pi_approx}.")
```

## We can repeat this process using nested loops to get more approximations and find the value of $\pi$ to a greater precision

```python
n_values = 10
for j in range(n_values):
    n_inside = 0
    for i in range(n_points):
        x = np.random.rand()
        y = np.random.rand()
        if (x**2 + y**2) < 1:
            n_inside += 1
    pi_approx = 4 * n_inside / n_points
```

```python
# Prepare an array to store the results
pi_values = np.empty(n_values)

# Calculate n_values separate estimate of pi, each using n_points random points
for j in range(n_values):
    n_inside = 0
    for i in range(n_points):
        x = np.random.rand()
        y = np.random.rand()
        if (x**2 + y**2) < 1:
            n_inside += 1
    pi_approx = 4 * n_inside / n_points
    pi_values[j] = pi_approx

print("Estimates of pi:")
print(pi_values)
```

```python
# This time generate a larger number of values
n_values = 1000

# Prepare an array to store the results
pi_values = np.empty(n_values)

# Calculate n_values separate estimate of pi, each using n_points random points
for j in range(n_values):
    n_inside = 0
    for i in range(n_points):
        x = np.random.rand()
        y = np.random.rand()
        if (x**2 + y**2) < 1:
            n_inside += 1
    pi_approx = 4 * n_inside / n_points
    pi_values[j] = pi_approx
```

## Plotting results in a histogram 
This is to see the most common value of $\pi$ produced.

```python
plt.rcParams['patch.force_edgecolor'] = True
plt.figure()
plt.hist(pi_values, bins=100, range=(2.9,3.4));

plt.xlabel(r'values of $\pi$')   
plt.ylabel('frequency')
plt.title(r'Histogram of most common $\pi$ values');
```

# Conclusion:
By using the Monte Carlo method to approximate $\pi$ by generating random numbers, we can see that the most frequent value of $\pi$ generated to 2dp is 3.14
